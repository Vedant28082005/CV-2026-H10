```python
def add_salt_pepper(img , prob = 0.3):
  noisy = img.copy()

  random = np.random.rand(*noisy.shape)
  noisy[random < prob/2] = 0
  noisy[random > 1-prob/2] = 255

  return noisy
```

```python
noisy = add_salt_pepper(img , 0.1)
plt.imshow(noisy , cmap='gray')
```

```python
plt.imshow(img , cmap='gray')
```

```python
img_wa = noisy.copy()

for i in range (img_wa.shape[0]):
  for j in range (img_wa.shape[1]):
    if(i+3<img_wa.shape[0] and j+3<img_wa.shape[1]):
      patch = img[i:i+3,j:j+3]
      img_wa[i,j] = np.average(patch)



plt.imshow(img_wa , cmap='gray')
```

```python
import numpy as np

arr = np.array([[1,2,4,8],[9,4,61,4]])

def _median(arr):
    arr = arr.reshape(-1)
    arr.sort()

    n = arr.size

    if n % 2 == 0:
        med = (arr[n//2 - 1] + arr[n//2]) / 2
    else:
        med = arr[n//2]

    return med

print(_median(arr))
```

```python
img_wme = noisy.copy()

for i in range (img_wme.shape[0]):
  for j in range (img_wme.shape[1]):
    if(i+3<img_wme.shape[0] and j+3<img_wme.shape[1]):
      patch = img[i:i+3,j:j+3]
      img_wme[i,j] = _median(patch)



plt.imshow(img_wme , cmap='gray')
```

```python
img_wmi = noisy.copy()

for i in range (img_wmi.shape[0]):
  for j in range (img_wmi.shape[1]):
    if(i+3<img_wmi.shape[0] and j+3<img_wmi.shape[1]):
      patch = img[i:i+3,j:j+3]
      img_wmi[i,j] = np.min(patch)



#plt.imshow(img_wmi , cmap='gray')
```

```python
img_wma = noisy.copy()

for i in range (img_wma.shape[0]):
  for j in range (img_wma.shape[1]):
    if(i+3<img_wma.shape[0] and j+3<img_wma.shape[1]):
      patch = img[i:i+3,j:j+3]
      img_wma[i,j] = np.max(patch)



#plt.imshow(img_wma , cmap='gray')
```

```python
images = [sample, img, noisy, img_wme, img_wa, img_wma, img_wmi]
titles = ["Original", "Grayscale", "Salt & Pepper", "Median", "Mean", "Max", "Min"]

fig, axes = plt.subplots(1, len(images), figsize=(4 * len(images), 4))

for ax, im, title in zip(axes, images, titles):
    if im.ndim == 2:  # grayscale image
        ax.imshow(im, cmap="gray")
    else:
        ax.imshow(im[..., ::-1])
    ax.set_title(title)
    ax.axis("off")

plt.tight_layout()
plt.show()
```
