```python
import numpy as np
import cv2
from google.colab.patches import cv2_imshow
import matplotlib.pyplot as plt
import scipy.stats as stats
```

```python
img = cv2.imread("/content/sample.png" , cv2.IMREAD_GRAYSCALE)
```

```python
plt.imshow(img , cmap='gray')
```

```python
def add_salt_pepper(img , prob = 0.3):
  noisy = img.copy()

  random = np.random.rand(*noisy.shape)
  noisy[random < prob/2] = 0
  noisy[random > 1-prob/2] = 255

  return noisy
```

```python
noisy = add_salt_pepper(img , 0.1)
plt.imshow(noisy , cmap='gray')
```

```python
plt.imshow(img , cmap='gray')
```

```python
img_w = img.copy()

for i in range (img_w.shape[0]):
  for j in range (img_w.shape[1]):
    if(i+3<img_w.shape[0] and j+3<img_w.shape[1]):
      patch = img[i:i+3,j:j+3]
      img_w[i,j] = np.average(patch)



plt.imshow(img_w , cmap='gray')
```

```python
img_w = img.copy()

for i in range (img_w.shape[0]):
  for j in range (img_w.shape[1]):
    if(i+3<img_w.shape[0] and j+3<img_w.shape[1]):
      patch = img[i:i+3,j:j+3]
      img_w[i,j] = np.median(patch)



plt.imshow(img_w , cmap='gray')
```

```python
img_w = img.copy()

for i in range (img_w.shape[0]):
  for j in range (img_w.shape[1]):
    if(i+3<img_w.shape[0] and j+3<img_w.shape[1]):
      patch = img[i:i+3,j:j+3]
      img_w[i,j] = np.min(patch)



plt.imshow(img_w , cmap='gray')
```
